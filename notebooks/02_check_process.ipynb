```python
import numpy as np
from pathlib import Path

# Find the repository root by looking for the dataset directory.
def find_repo_root(start: Path = Path.cwd()) -> Path:
    p = start.resolve()
    for candidate in [p] + list(p.parents):
        if (candidate / "dataset").exists():
            return candidate
    raise FileNotFoundError("Could not locate the repository root containing the dataset/ directory")

repo_root = find_repo_root()
print("repo_root=", repo_root)

spec_path = repo_root / "dataset" / "processed" / "logmel_128" / "acafly" / "XC11209.npy"
if not spec_path.exists():
    raise FileNotFoundError(f"Spec file not found: {spec_path}")

spec = np.load(spec_path)
print(spec.shape)
print(spec.dtype)
```

```python
import librosa.display
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

librosa.display.specshow(
    spec,
    x_axis="time",
    y_axis="mel",
    sr=32000
)

plt.colorbar(format="%+2.0f dB")
plt.title("Log-Mel Spectrogram")

plt.tight_layout()
plt.show()
```
