```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
```

```python
audio_path=Path("../dataset/raw/train_short_audio/acafly")
sample=list(audio_path.glob("*.ogg"))[0]
sample
```

```python
audio,sr=librosa.load(sample)
audio,sr
```

```python
plt.plot(audio)
```

```python
mel_spec=librosa.feature.melspectrogram(
    y=audio,
    sr=sr,
    n_fft=2048,
    n_mels=128
)
plt.plot(mel_spec)
plt.show()
```

```python
spec=librosa.power_to_db(mel_spec,ref=np.max)
librosa.display.specshow(spec, sr=sr, x_axis='time', y_axis='mel')
plt.colorbar(format='%+2.0f dB')

plt.title("Log-Mel Spectrogram")

plt.tight_layout()

plt.show()
```
